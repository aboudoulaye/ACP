```python
import pandas as pd

# Dataset 1 - Sleep Study
sleep_types = pd.DataFrame({
    'Colonne': ['Age', 'Sleep Duration', 'Gender', 'Enough Sleep'],
    'Type': ['Quantitative', 'Quantitative', 'Qualitative', 'Qualitative'],
    'Raison': [
        'Nombre mesurable',
        'Valeur numérique (heures)',
        'Catégorie (M/F)',
        'Réponse catégorielle (oui/non)'
    ]
})
print("=== Sleep Study ===")
print(sleep_types.to_string(index=False))

# Dataset 2 - Mental Health
mental_types = pd.DataFrame({
    'Colonne': ['Year', 'Country', 'Prevalence (%)', 'Disorder Type'],
    'Type': ['Quantitative', 'Qualitative', 'Quantitative', 'Qualitative'],
    'Raison': [
        'Valeur numérique',
        'Nom de pays (catégorie)',
        'Pourcentage mesurable',
        'Catégorie de maladie'
    ]
})
print("\n=== Mental Health ===")
print(mental_types.to_string(index=False))

# Dataset 3 - Credit Card
credit_types = pd.DataFrame({
    'Colonne': ['Income', 'Age', 'Approved', 'Employment Status'],
    'Type': ['Quantitative', 'Quantitative', 'Qualitative', 'Qualitative'],
    'Raison': [
        'Montant en chiffres',
        'Nombre mesurable',
        'Résultat catégoriel (oui/non)',
        'Catégorie (employé/chômeur)'
    ]
})
print("\n=== Credit Card ===")
print(credit_types.to_string(index=False))
```

```python
import pandas as pd
from sklearn.datasets import load_iris

# Charger le dataset Iris
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target_names[iris.target]

# Afficher les premières lignes
print(df.head())
print("\n")

# Identifier les types de colonnes
quantitatives = ['sepal length (cm)', 'sepal width (cm)',
                 'petal length (cm)', 'petal width (cm)']
qualitatives = ['species']

print("=== Colonnes QUANTITATIVES ===")
for col in quantitatives:
    print(f"- {col} : mesure en cm → valeur numérique")

print("\n=== Colonnes QUALITATIVES ===")
for col in qualitatives:
    print(f"- {col} : nom de l'espèce → catégorie")
```
